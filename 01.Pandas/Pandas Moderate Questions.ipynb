<span style='color:orange'>

Round the output to 1 decimal places, and append a percent sign (%) to the end (Ana doesn't want to have to do that later manually each time - I know....).

## 12. Multi-Level Marketing

<div style='background-color:darkblue; padding:15px;'>

<span style='color:orange'>
    
Sally is your average Stay at Home Mom so she decided to join an MLM (Multi Level Marketing) company.

She wants to **look at the months where she made a profit during the first 6 months of the year** (`January - June`).

Write a query to return the Month (number) and Profit that was made for each of those months.

**Order output by profit highest to lowest.**

Note: Any negative number in the "Profit" column are expenses for running the business.




</div>

---

```python
profits = pd.read_excel('datasets/Questions/Moderate/12.Multi-Level Marketing.xlsx')
profits
```

```python
profits['Month'] = profits['date'].dt.month
profits
```

```python
profits.groupby('Month')['profit'].sum().reset_index()
```

```python
df_group = profits.groupby(by='Month')['profit'].sum().reset_index(name='agg_profits')
df_group
```

```python
result = df_group[(df_group['Month'] <= 6) & (df_group['agg_profits'] > 0)].sort_values(by='agg_profits', ascending=False)
result
```

---
---

```python
# ----------------------------------------Teachers Solution----------------------------------------
profits['date'] = pd.to_datetime(profits['date'])  
df_filtered = profits[profits['date'].dt.year == 2024]
df_filtered = df_filtered[df_filtered['date'].dt.month.between(1, 6)]

agg_profits = df_filtered.groupby(df_filtered['date'].dt.month)['profit'].sum().reset_index(name='agg_profits')
agg_profits = agg_profits[agg_profits['agg_profits'] > 0]
agg_profits.sort_values(by='agg_profits', ascending=False)
```

The Capital gain/loss of a stock is the total gain or loss after buying and selling the stock one or many times.

Provide the stock name and Capital gain/loss as "capital_gain_loss".

</div>

---

```python
df_33 = pd.read_excel('datasets/Questions/Moderate/33.Stock_Price.xlsx')
df_33.head()
```

```python
df_buy = df_33[df_33['action'] == 'Buy'].groupby('stock')['price'].sum().reset_index()
df_buy
```

```python
df_sell = df_33[df_33['action'] == 'Sell'].groupby('stock')['price'].sum().reset_index()
df_sell
```

```python
df_merged = df_sell.merge(df_buy, on='stock', how='left', suffixes=('_sell', '_buy'))
df_merged['Total'] = df_merged['price_sell'] - df_merged['price_buy']
df_merged
```

```python
result = df_merged[['stock', 'Total']]
result.columns =['Stock', 'Capital_Gain_Loss']
result
```
