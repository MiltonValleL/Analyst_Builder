# 4. REORDERING AND SORTING INDEXES

</span>

---

```python
df = pd.read_csv('datasets/WorldContinents.csv', encoding='ISO 8859-1')
df
```

```python
df.set_index(['continent']).sort_index()
```

```python
pd.set_option('display.max_rows', 250)

# this line of code will sort both sub-indexes...
df.set_index(['continent', 'sub_region']).sort_index()
```

```python
# This line of code will sort only the 'sub_region'...
df.set_index(['continent', 'sub_region']).sort_index(level='sub_region') ### Take a look at this line
```

```python
# Another version of the previous code....

df.set_index(['continent', 'sub_region']).sort_index(level='sub_region', ascending=False)
```

```python
# Another version would be...

df.set_index(['continent', 'sub_region']).sort_index(ascending=[True, False])
```

```python
# Assigning the changes to the main DataFrame...

df.set_index(['continent', 'sub_region'], inplace=True)
df
```

```python
# Now we will swap both indexes...

df.swaplevel('continent', 'sub_region')
```

```python
# The original DataFrame, didn't change
df
```

```python
df = df.swaplevel('continent', 'sub_region')
df
```
