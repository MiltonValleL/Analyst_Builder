```python
age_group = food['Age'].value_counts(ascending=False).reset_index()
age_group
```

```python
plt.figure(figsize=(12, 6))
plt.bar(x=age_group['Age'], height=age_group['count'])
plt.xticks(list(range(20, 81, 5)))
plt.show()
```

```python
age_groups = [(23, 30, 'Group 1'), (31, 40, 'Group 2'), (41, 50, 'Group 3'),
              (51, 60, 'Group 4'), (61, 70, 'Group 5'), (71, 85, 'Group 6')]

def assign_age_group(Age):
    for age_range in age_groups:
        if age_range[0] <= Age <= age_range[1]:
            return age_range[2]
    return 'Unknown'

food['Age_Group'] = food['Age'].apply(assign_age_group)
```

```python
food
```

```python
age_groups2 = [(23, 30), (31, 40), (41, 50),
              (51, 60), (61, 70), (71, 85)]

def assign_age_group_2(Age):
    for age_range in age_groups:
        if age_range[0] <= Age <= age_range[1]:
            return f"{age_range[0]} - {age_range[1]}"
    return 'Unknown'

food['Age_Group2'] = food['Age'].apply(assign_age_group_2)
```

```python
food[['Age', 'Age_Group2', 'Age_Group']].sample(9)
```

```python
age_order = ['23 - 30', '31 - 40', '41 - 50', '51 - 60', '61 - 70', '71 - 85']

sns.pointplot(data=food, x='Age_Group2', y='AcceptedCmpOverall', order=age_order)
```

```python
age_counts = food['Age_Group2'].value_counts()
age_counts
```

```python
age_percent = (age_counts / food.shape[0]).reset_index()
age_percent.columns = ['age_group', 'percentage']
age_percent = age_percent.sort_values('age_group')
age_percent
```

```python
sns.barplot(data=age_percent, x='age_group', y='percentage', hue='age_group')
plt.title('Percentage of Accepted Campaigns per Age Group')
plt.show()
```

---

<span style='color:orange'>
    
**AGE SEGMENTATION:** According to the upper graph, we can conclude that people aged 23 to 30 and 71 to 85 are the least represented in this group. They reach, at most, 7.5% of the total number of customers that UFOOD company has.

On the other hand, the most relevant age group comprises people aged 41 to 50, as they represent 33% of the whole group. In second place is the group of people aged 51 to 60, representing 22.5% of the total.
