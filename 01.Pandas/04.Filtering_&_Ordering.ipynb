# 3. FILTERING USING STRING METHODS

</span>

---

---
<div style='background-color:lightgray; padding:15px'>


<span style='color:green'>

1. `.contains()` method is used to check if each element in a Series contains a specified string or pattern. This method is particularly useful when you want to filter rows based on whether a specific substring exists in the values of a column.


2. `.str.startswith()` method is used to check if each element in a Series starts with a specified string.


3. `.str.endswith()` method is used to check if each element in a Series ends with a specified string.


4. `.duplicated()` method is used to identify duplicate rows in a DataFrame


5. `.str.match()` method is used to check if each element in a Series matches a specified regular expression pattern

</div>

---

<span style='color:green'>

## 3.1. Method `.contains()`

</span>

```python
df[df['Product'].str.contains('MacBook')]
```

```python
df[~df['Product'].str.contains('MacBook')]
```

```python
df[df['Product'].str.contains('MacBook') == False]
```

```python
df[(df['Product'].str.contains('MacBook')) & (df['Memory'].str.contains('SSD'))]
```

## 3.5. Filtering Using Date Functions

</span>

---

<div style='background-color:lightgray; padding:15px'>

<span style='color:green'>
    
These functions are extremely useful to manipulate and use date columns within a Pandas Dataframe. Here are some of the most common ones you'll use:

  -  **`pd.to_datetime()`** function is used to convert an object to a datetime object.

  -  **`.dt.year`** is used to extract the year component from a datetime-like Series.

  -  **`.dt.day`** is used to extract the day component.

  -  **`.dt.day_name()`** is used to extract the name of the day of the week and is particularly useful when you want to analyze or categorize data based on the day of the week.


</div>

---

```python
df = pd.read_csv('datasets/WeatherHistory.csv')
df
```

```python
df.info()
```

```python
df['Formatted Date'] = pd.to_datetime(df['Formatted Date'])
df
```

```python
df.info()
```

```python
df[df['Formatted Date'].dt.year == 2006]
```

```python
df[df['Formatted Date'].dt.month == 3]
```

```python
df[df['Formatted Date'].dt.day == 26]
```

```python
# VERY IMPORTANT TO REMEMBER!!!
# This line of code shows us how to filter by Day of the Week.

df[df['Formatted Date'].dt.day_name() == 'Monday']
```

# 5. Ordering DataFrames

</span>

---

<div style='background-color:lightgray; padding:15px;'>

<span style='color:green'>
    
   * **<span style='color:brown'>.sort_values()</span>** method in pandas is used to sort a DataFrame or Series by the values in one or more columns

Basic Syntax:

`df.sort_values(by, axis=0, ascending=True, inplace=False, kind='quicksort', na_position='last', ignore_index=False, key=None)`

</span>

</div>

```python
df = pd.read_csv('datasets/laptop_prices.csv', encoding='ISO-8859-1')
df
```

```python
df.filter(['Company', 'Product', 'Cpu', 'Ram', 'Memory', 'Gpu', 'Price_usd']).sort_values(by='Price_usd')
```

```python
df.filter(['Company', 'Product', 'Cpu', 'Ram', 'Gpu', 'Price_usd']).sort_values(by='Price_usd',
                                                                               ascending=False)
```

```python
df[df['Product'] == 'Blade Pro'].sort_values(by='Price_usd', ascending=False)
```

```python
df[df['Company'] == 'Razer'].sort_values(['Inches', 'Price_usd'], ascending=[False, True])
```

```python
df[df['Company'] == 'Razer'].sort_values(['Inches', 'Price_usd'], ascending=[True, False])
```
