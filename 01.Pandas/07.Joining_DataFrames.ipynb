# 2. BASICS OF MERGE AND JOIN



```python
import pandas as pd
```

```python
data1 = {
    'employee_id': [1, 2, 3, 4, 5, 6, 7, 8],
    'name': ['Michael Scott', 'Jim Halpert', 'Pam Beesly', 'Dwight Schrute', 'Angela Martin', 'Kevin Malone', 'Oscar Martinez', 'Stanley Hudson'],
    'job_title': ['Regional Manager', 'Sales Representative', 'Receptionist', 'Assistant to the Regional Manager', 'Head of Accounting', 'Accountant', 'Accountant', 'Sales Representative'],
    'salary': [100000, 60000, 55000, 65000, 60000, 55000, 55000, 60000]
}

df1 = pd.DataFrame(data1)

data2 = {
    'employee_id': [1, 2, 3, 5, 6, 7, 8, 9],
    'name': ['Michael Scott', 'Jim Halpert', 'Pam Beesly', 'Angela Martin', 'Kevin Malone', 'Oscar Martinez', 'Stanley Hudson', 'Phyllis Vance'],
    'department': ['Management', 'Sales', 'Reception', 'Accounting', 'Accounting', 'Accounting', 'Sales', 'Sales']
}

df2 = pd.DataFrame(data2)
```

```python
df1
```

```python
df2
```

## 2.1. Use of `merge()` Function

</span>

```python
df1.merge(df2, how='left')
```

```python
df1.merge(df2, how='right')
```

```python
df1.merge(df2, how='outer')
```

```python
df1.merge(df2, how='outer', on='employee_id')
```

```python
df1.merge(df2, how='outer', on=['employee_id', 'name'])
```

```python
data3 = {
    'emp_id': [1, 2, 3, 5, 6, 7, 8, 9],
    'name': ['Michael Scott', 'Jim Halpert', 'Pam Beesly', 'Angela Martin', 'Kevin Malone', 'Oscar Martinez', 'Stanley Hudson', 'Phyllis Vance'],
    'department': ['Management', 'Sales', 'Reception', 'Accounting', 'Accounting', 'Accounting', 'Sales', 'Sales']
}

df3 = pd.DataFrame(data3)
df3
```

```python
df1.merge(df3, how='outer', left_on='employee_id', right_on='emp_id')
```

```python
# It is not possible to use in the same line of code the parameters 'on' and 'left_on' with 'right_on'
# This causes a 'MergeError', as you can see below...

df1.merge(df3, how='outer', on='name', left_on='employee_id', right_on='emp_id')
```

# 3. MERGE OPERATIONS



```python
data1 = {
    'employee_id': [1, 2, 3, 4, 5, 6, 7, 8],
    'name': ['Michael Scott', 'Jim Halpert', 'Pam Beesly', 'Dwight Schrute', 'Angela Martin', 'Kevin Malone', 'Oscar Martinez', 'Stanley Hudson'],
    'job_title': ['Regional Manager', 'Sales Representative', 'Receptionist', 'Assistant to the Regional Manager', 'Head of Accounting', 'Accountant', 'Accountant', 'Sales Representative'],
    'salary': [100000, 60000, 55000, 65000, 60000, 55000, 55000, 60000]
}

df1 = pd.DataFrame(data1)

data2 = {
    'employee_id': [1, 2, 3, 5, 6, 7, 8, 9],
    'name': ['Michael Scott', 'Jim Halpert', 'Pam Beesly', 'Angela Martin', 'Kevin Malone', 'Oscar Martinez', 'Stanley Hudson', 'Phyllis Vance'],
    'department': ['Management', 'Sales', 'Reception', 'Accounting', 'Accounting', 'Accounting', 'Sales', 'Sales']
}

df2 = pd.DataFrame(data2)


data3 = {
    'employee_id': [1, 2, 3, 4, 5, 6, 7, 8],
    'name': ['Michael Scott', 'Creed Bratton', 'Pam Beesly', 'Dwight Schrute', 'Angela Martin', 'Kevin Malone', 'Oscar Martinez', 'Stanley Hudson'],
    'job_title': ['Regional Manager', 'Sales Representative', 'Receptionist', 'Assistant to the Regional Manager', 'Head of Accounting', 'Accountant', 'Accountant', 'Sales Representative'],
    'salary': [100000, 60000, 55000, 65000, 60000, 55000, 55000, 60000]
}

df3 = pd.DataFrame(data3)
```

```python
pd.merge(df1, df2, on='employee_id', how='inner')
```

```python
# One way to avoid the 'name_x' and 'name_y' columns is this...

pd.merge(df1, df2, on=['employee_id', 'name'], how='outer')
```

```python
# Another way to do the same is...

pd.merge(df1, df2[['employee_id', 'department']], on='employee_id', how='outer')
```

```python
# And finally one last way to do the same is...

pd.merge(df1[['employee_id', 'job_title', 'salary']], df2, on='employee_id', how='outer')
```

```python
pd.merge(df1, df2, on='employee_id', how='outer', suffixes=('_1', '_2'))
```

```python
df1.merge(df2, on='employee_id', how='outer', suffixes=('_1', '_2'), indicator=True)
```

---

```python
df2.head()
```

```python
df3.head()
```

```python
df3.merge(df2, on='employee_id', how='outer', indicator=True)
```

---

<span style='color:orange'>Take a look a the index '1'. That row has to different names on it.

---

```python
# With this line of code the incongruency of the name in the row No. 2 has dissapired. 
# There is no 'Creed Bratton' and there is no 'Jim Halpert' either...

pd.merge(df3, df2, on=['employee_id', 'name'], how='inner')
```

```python
# But if we use the parameter "outer" on 'how', now both names appear. 
# Nevertheless the "employee_id" No.2 is assigned to both names...

pd.merge(df3, df2, on=['employee_id', 'name'], how='outer')
```

# 5. MERGING MULTIPLE DATAFRAMES


```python
data1 = {
    'employee_id': [1, 2, 3, 4, 5, 6, 7, 8],
    'name': ['Michael Scott', 'Jim Halpert', 'Pam Beesly', 'Dwight Schrute', 'Angela Martin', 'Kevin Malone', 'Oscar Martinez', 'Stanley Hudson'],
    'job_title': ['Regional Manager', 'Sales Representative', 'Receptionist', 'Assistant to the Regional Manager', 'Head of Accounting', 'Accountant', 'Accountant', 'Sales Representative'],
    'salary': [100000, 60000, 55000, 65000, 60000, 55000, 55000, 60000]}
demographics = pd.DataFrame(data1)


data2 = {
    'employee_id': [1, 2, 3, 5, 6, 7, 8, 9],
    'name': ['Michael Scott', 'Jim Halpert', 'Pam Beesly', 'Angela Martin', 'Kevin Malone', 'Oscar Martinez', 'Stanley Hudson', 'Phyllis Vance'],
    'department_id': [1, 2, 3, 4, 4, 4, 2, 2]}
name_dep = pd.DataFrame(data2)


data3 = {
    'employee_id': [1, 2, 3, 4, 5, 6, 7, 8, 9],
    'manager_id': [None, 1, 1, 1, 1, 4, 4, 1, 1]}
manager = pd.DataFrame(data3)


data4 = {
    'department_id': [1, 2, 3, 4],
    'department': ['Management', 'Sales', 'Reception', 'Accounting']}
department = pd.DataFrame(data4)
```

```python
# I will take a look first to the 'manager'DF
manager
```

```python
# neet to make changes on the 'manager_id' column...
manager['manager_id'] = manager['manager_id'].astype('Int64')
manager
```

```python
# Now I will take a look at 'department' DF
department
```

```python
name_dep
```

```python
demographics
```

```python
final_df = demographics.merge(name_dep, 
                              on=['employee_id', 'name'], 
                              how='outer').merge(department, on='department_id', 
                                                 how='outer').merge(manager, on='employee_id', how='outer')

final_df
```

---

```python
final_df['department_id'] = final_df['department_id'].astype('Int64')
final_df
```

# 6. CONCATENATION


```python
data1 = {
    'employee_id': [1, 2, 3, 4, 5, 6, 7, 8],
    'name': ['Michael Scott', 'Jim Halpert', 'Pam Beesly', 'Dwight Schrute', 'Angela Martin', 'Kevin Malone', 'Oscar Martinez', 'Stanley Hudson'],
    'job_title': ['Regional Manager', 'Sales Representative', 'Receptionist', 'Assistant to the Regional Manager', 'Head of Accounting', 'Accountant', 'Accountant', 'Sales Representative'],
    'salary': [100000, 60000, 55000, 65000, 60000, 55000, 55000, 60000]}
df1 = pd.DataFrame(data1)

data2 = {
    'employee_id': [1, 2, 3, 5, 6, 7, 8, 9],
    'name': ['Michael Scott', 'Jim Halpert', 'Pam Beesly', 'Angela Martin', 'Kevin Malone', 'Oscar Martinez', 'Stanley Hudson', 'Phyllis Vance'],
    'department': ['Management', 'Sales', 'Reception', 'Accounting', 'Accounting', 'Accounting', 'Sales', 'Sales']}
df2 = pd.DataFrame(data2)
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1, df2])
```

```python
pd.concat([df1, df2], axis=1)
```

```python
pd.concat([df1.set_index('employee_id'), df2.set_index('employee_id')], axis=1)
```

```python
edit_df = pd.concat([df1[['employee_id', 'name']], df2[['employee_id', 'name']]], axis=0)
edit_df
```

```python
# Now I will drop duplicated data...

edit_df.drop_duplicates(inplace=True)
edit_df
```

```python
# now it's necessary to reset the index...
edit_df.reset_index(inplace=True, drop=True)
edit_df
```

```python
ex2 = pd.concat([df1, df2], axis=0)
ex2
```

```python
ex3 = pd.concat([df1, df2], axis=0, ignore_index=True)
ex3
```

```python
ex4 = pd.concat([df1, df2], keys=['DF1', 'DF2'])
ex4
```
