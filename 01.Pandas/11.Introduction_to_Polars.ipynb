```python
import polars as pl
```

```python
insurance = pl.read_csv('datasets/health_insurance.csv')
insurance.head()
```

```python
insurance.select(pl.col('smoker'))
```

```python
insurance.select(pl.col('smoker', 'sex', 'age'))
```

```python
insurance.select(pl.col('*'))
```

```python
insurance_month = insurance.with_columns((pl.col('charges')/12).alias('charges_per_month'))
insurance_month
```

```python
insurance_month.filter(pl.col('age') >= 30)
```

```python
insurance_month.filter((pl.col('age') >= 30) & (pl.col('sex') == 'male')) 
```

```python
insurance_month.group_by('sex').mean()
```

```python
insurance_month.group_by('sex').len()
```

```python
insurance_month.group_by('sex').sum()
```
