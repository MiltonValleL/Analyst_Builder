## 1.1. Series
    
</span>

---

<div style='background-color:lightgray; padding:15px;'>

<span style='color:green'>

A Series is essentially a one-dimensional labeled array that can hold data of any type (integers, floats, strings, etc.). It is similar to a column in a spreadsheet or a single column of data in a database table. A Series has two main components:

 - **Data**: This is a NumPy array that holds the actual values or elements of the Series.
 
 
 - **Index**: This is a label or identifier for each element in the Series. By default, it is a sequence of integers from 0 to N-1, where N is the length of the Series. However, you can customize the index to be any set of labels, such as strings or dates, making Series highly versatile.

</div>

---

---

<span style='color:red'>
    
## 1.2. DataFrame
    
</span>

---

<div style='background-color:lightgray; padding:15px;'>

<span style='color:green'>

A DataFrame is a two-dimensional tabular data structure in pandas. It is similar to a spreadsheet or a SQL table. It consists of rows and columns, where each column is a Series. DataFrames are incredibly powerful and flexible for handling and analyzing structured data. Key features of a DataFrame include:

 - **Rows and Columns**: Rows represent individual records or observations, while columns represent variables or attributes.

 - **Index**: DataFrames have both row and column indexes, which can be customized. The column index is also known as column labels.

 - **Heterogeneous Data**: Each column in a DataFrame can contain data of different types, including integers, floats, strings, or even other complex data structures.

</div>

## 2.1. NumPy Array
    
</span>

---

<div style='background-color:lightgray; padding:15px;'>

<span style='color:green'>
    
 * NumPy arrays are homogeneous, which means they can only store data of a single data type (e.g., integers, floats). This uniformity in data type allows for more efficient computation.

 * NumPy arrays are essential for performing mathematical and logical operations on large datasets and are commonly used in scientific computing, data analysis, and machine learning.

 * You can create a NumPy array from a Python list or other iterable.

</div>

## 2.2. Pandas Series
    
</span>

---

<div style='background-color:lightgray; padding:15px;'>

<span style='color:green'>
    
 * A pandas Series is essentially a one-dimensional array, similar to a NumPy array. However, it includes an index that labels each element in the Series, making it possible to access data using these labels.

 * Pandas Series can store data of different types (*integers, floats, strings*) within the same Series. This allows for greater flexibility when dealing with real-world datasets.

 * Series are commonly used in data preprocessing and analysis, as well as in time series data, and they can be used as columns in pandas DataFrames.

</div>

## 2.3. Pandas DataFrames
    
</span>

---


```python
import pandas as pd

index_values = ['Best', '2nd Best', '3rd Best']
my_list = [10, 25, 35]

one = pd.Series(data=my_list, index=index_values)
one
```

```python
index_values2 = ['Best', '2nd Best', '4th Best']
my_list2 = [10, 25, 40]

two = pd.Series(data=my_list2, index=index_values2)
two
```

```python
df = pd.DataFrame(data={'Series 1':one, 'Series 2':two})
df
```

```python
ice_cream = [['MCC', 1], ['Butterscoth', 2], ['Butter Pecan', 3]]
```

```python
pd.DataFrame(data=ice_cream, index=['Flavor 1', 'Flavor 2', 'Flavor 3'], columns=('Flavor', 'Scoops'))
```

```python
pd.DataFrame(data=ice_cream, columns=('Flavor', 'Scoops'))
```

```python
pd.DataFrame(data=ice_cream)
```
