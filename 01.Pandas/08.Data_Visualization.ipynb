# 3. INTRODUCTION TO MATPLOTLIB


---
---

<span style='color:red'>

## 3.1. Example No. 1


```python
x = [1, 2, 3]
y = [5, 10, 20]

plt.plot(x, y, label='First Label', color='red', linestyle='--', marker='o', markersize=18, markeredgecolor='blue')

plt.title('Matplotlib Viz', fontdict={'fontname':'Times New Roman', 'fontsize':30, 'color':'brown'})
plt.xlabel('X Axis', fontdict={'fontname':'Arial', 'fontsize':15, 'color':'green'})
plt.ylabel('Y Axis', fontdict={'fontname':'Courier New', 'fontsize':18, 'color':'blue'})

plt.xticks([1, 2, 3])
plt.yticks([0, 5, 10, 15, 20])

plt.legend()
plt.show()
```

---
---

<span style='color:red'>

## 3.2. Example No. 2


```python
x = [1, 4, 8]
y = [2, 20, 93]
z = [1, 12, 75]

plt.plot(x, y, label='First Label', color='blue', linestyle='-.', marker='^', markersize=12, markeredgecolor='red')
plt.plot(x, z, label='Second Label', color='green', linestyle='--', marker='s', markersize=12, markeredgecolor='orange')

plt.title('Second Matplotlib Viz', fontdict={'fontname':'Courier New', 'fontsize':27, 'color':'brown'})
plt.xlabel('X Axis', fontdict={'fontname':'Courier New', 'fontsize':15, 'color':'darkred'})
plt.ylabel('Y Axis', fontdict={'fontname':'Courier New', 'fontsize':15, 'color':'darkred'})

plt.xticks(list(range(1,10,1)))
plt.yticks(list(range(0,100,10)))

plt.legend()

plt.show()
```

---
---

<span style='color:red'>

## 3.3. Example No. 3


```python
x = [1, 4, 8]
y = [2, 20, 93]
z = [1, 12, 75]

plt.figure(figsize=(12,6))

plt.plot(x, y, label='First Label', color='blue', linestyle='-.', marker='^', markersize=12, markeredgecolor='red')
plt.plot(x, z, label='Second Label', color='green', linestyle='--', marker='s', markersize=12, markeredgecolor='orange')

plt.title('Second Matplotlib Viz', fontdict={'fontname':'Courier New', 'fontsize':27, 'color':'brown'})
plt.xlabel('X Axis', fontdict={'fontname':'Courier New', 'fontsize':15, 'color':'darkred'})
plt.ylabel('Y Axis', fontdict={'fontname':'Courier New', 'fontsize':15, 'color':'darkred'})

plt.xticks(list(range(0,10,1)))
plt.yticks(list(range(0,100,10)))

plt.legend()

plt.savefig('matplotlib.png')
plt.show()
```

# 5. ADVANCED DATA VISUALIZATION


```python
df = pd.read_csv('datasets/smoking.csv')
df
```

```python
data1 = df.groupby(by=['gender', 'age'])['smoke'].count()
data1
```

```python
smokers = df[df['smoke'] == 'Yes']
female_smokers = smokers[smokers['gender'] == 'Female']
male_smokers = smokers[smokers['gender'] == 'Male']
```

```python
# Since here ara three datasets of smokers we will create subplots...
fig, axs = plt.subplots(1, 2)

sns.histplot(female_smokers['age'], bins=25, ax=axs[0], color='blue') 
axs[0].set_title('Age Distribution for Females')

sns.histplot(male_smokers['age'], bins=25, ax=axs[1], color='orange')
axs[1].set_title('Age Distribution for Males')
plt.show()
```

```python
fig, axs = plt.subplots(1, 2)

sns.histplot(female_smokers['age'], bins=50, ax=axs[0], color='orange')
axs[0].set_title('Age Distribution for Females')

sns.histplot(male_smokers['age'], bins=50, ax=axs[1], color='green')
axs[1].set_title('Age Distribution for Males')

plt.tight_layout(pad=1.2)
plt.show()
```

---

```python
housing = pd.read_csv('datasets/real_estate.csv')
housing
```

```python
housing.corr(numeric_only=True)
```

```python
housing.drop(labels='zip_code', axis=1, inplace=True)
housing
```

```python
correlation = housing.corr(numeric_only=True)
correlation
```

---
---

<span style='color:red'>

## 5.2 Pearson Correlation (*Default Setting*)


```python
sns.heatmap(correlation, annot=True, cmap='coolwarm')
```

---

<span style='color:orange'>**The PEARSON CORRELATION assumes that the data is `normaly distribuited`. So one can conclude that this correlation is `very optimistic`.**

---

```python
pearson_correlation = housing.corr(method='pearson', numeric_only=True)

sns.heatmap(pearson_correlation, annot=True, cmap='coolwarm')
plt.title('Pearson Correlation')
plt.show()
```

---
---

<span style='color:red'>

## 5.4 Spearman Correlation 


---

<span style='color:orange'>**The SPEARMAN CORRELATION uses a `Monotonic Relationship` between two variables. So with the SPEARMAN CORRELATION `we are not going to take the outliers into as much consideration` rather than the core of the data.**

---

```python
spearman_corr = housing.corr(method='spearman', numeric_only=True)

sns.heatmap(spearman_corr, annot=True, cmap='coolwarm')
plt.title('Spearman Correlation')
plt.show()
```

---
---

<span style='color:red'>

## 5.3 Kendall Correlation 


---

<span style='color:orange'>**The KENDALL CORRELATION is even more strict than the SPEARMAN CORRELATION in terms of outliers `it's gonna weigh those outliers even a little bit less` so when we use the KENDALL CORRELATION we will notice that the numbers of the correlation matrix are even smaller then the SPEARMAN CORRELATION.**

---

```python
kendall_corr = housing.corr(method='kendall', numeric_only=True)

sns.heatmap(kendall_corr, annot=True, cmap='coolwarm')
plt.title('Kendall Correlation')
plt.show()
```

---
---

<span style='color:red'>

## 5.4 Sum of Correlation Plots

```python
fig, axs = plt.subplots(3,1, figsize=(6, 9))

sns.heatmap(pearson_correlation, annot=True, cmap='coolwarm', ax=axs[0])
axs[0].set_title('Pearson Correlation')

sns.heatmap(spearman_corr, annot=True, cmap='coolwarm', ax=axs[1])
axs[1].set_title('Spearman Correlation')

sns.heatmap(kendall_corr, annot=True, cmap='coolwarm', ax=axs[2])
axs[2].set_title('Kendall Correlation')

plt.tight_layout()
plt.show()
```
