## 1.1. Grouping in Pandas

</span>



---

<div style='background-color:lightgray; padding:15px;'>

<span style='color:green'>

1. **Definition:**

    * Grouping involves dividing a DataFrame into groups based on values in one or more columns.

2. **Key Components:**

    * **Grouping Column(s):** The column(s) used to define the groups.
  
    * **GroupBy Object:** The result of the grouping operation, representing the grouped data.

3. **GroupBy Object:**

    * Acts as an intermediary that allows you to apply various aggregation functions on each group.

</div>

---

---
---

<span style='color:red'>
    
## 1.2. Aggregations in Pandas

</span>

---

<div style='background-color:lightgray; padding:15px;'>

<span style='color:green'>
    
1. **Definition:**

    * Aggregations involve performing a calculation on a group of data points, resulting in a single aggregated value.


2. **Common Aggregation Functions:**

    * **Sum:** Computes the sum of values in a group.
  
    * **Mean (Average):** Computes the mean or average of values in a group.
  
    * **Count:** Counts the number of non-null values in a group.
  
    * **Min and Max:** Find the minimum and maximum values in a group.
  
    * **Custom Functions:** You can apply custom aggregation functions.
  


</div>

# 2. AGGREGATIONS WITH GROUP BY

---

```python
df = pd.read_csv('datasets/banking.csv')
df.info()
```

```python
df
```

```python
df.groupby(by='education')['outstanding_balance'].min()
```

```python
df.groupby(by='education')['outstanding_balance'].max()
```

```python
df.groupby(by='education')['outstanding_balance'].mean()
```

```python
df.groupby(by='education')['outstanding_balance'].sum().sort_values(ascending=False)
```

```python
df.groupby(by='education')['outstanding_balance'].count().sort_values(ascending=False)
```

```python
df.groupby(by='education')['outstanding_balance'].median().sort_values(ascending=False)
```

```python
output = df.groupby(by='education').agg({'outstanding_balance':['sum', 'count', 'mean', 'median', 'min', 'max']})
output
```

---

<span style='color:orange'>

**SOMETHING TO REMEMBER:**

TAKE A LOOK AT THE CALCULATION FORMAT OF THE "AVERAGE" COLUMN. **`IT'S SOMETHING NEW TO LEARN`**

---

```python
output['average'] = ((output['outstanding_balance']['sum']) / (output['outstanding_balance']['count'])) * 100
output.sort_values(by='average', ascending=False)
```

---

<span style='color:orange'>

**SOMETHING TO REMEMBER:**

TAKE A LOOK AT THE RENAMING OF THE COLUMNS FORMAT. **`IT'S SOMETHING NEW TO LEARN TOO.`**

---

```python
df.groupby(by='education').agg(sum_balance = ('outstanding_balance', 'sum'), count_balance = ('outstanding_balance', 'count'))
```

# 3. GROUP BY ON MULTIPLE COLUMNS

---

```python
df = pd.read_csv('datasets/banking.csv')
df
```

```python
df.groupby(by=['job', 'marital'])['outstanding_balance'].sum()
```

```python
df.groupby(by=['job', 'marital'])['outstanding_balance'].sum().reset_index()
```

```python
df[df['job'] == 'management'].groupby(by=['job', 'marital'])['outstanding_balance'].sum().reset_index()
```

```python
df.groupby(by=['education', 'personal_loan'])['outstanding_balance'].mean()
```

```python
new_df = df.groupby(by=['education', 'personal_loan'])['outstanding_balance'].mean().reset_index()
new_df
```

# 4. TRANSFORM WITH GROUP BY **

---

```python
df = pd.read_csv('datasets/banking.csv')
```

```python
sub_df = df[:500]
sub_df
```

---

<span style='color:orange'>

**SOMETHING TO REMEMBER:**

Using the TRANSFORM method, we can fill a new column with the information generated with the "groupby", take a look in the cell behind

</span>

---

```python
sub_df['avergage_balance_per_job'] = sub_df.groupby(by='job')['outstanding_balance'].transform('mean')
sub_df
```

```python
df = pd.read_csv('datasets/banking.csv')

new_df = df[:500]

new_df['avg_balance_per_job'] = new_df.groupby(by='job')['outstanding_balance'].transform('mean')
new_df
```

---

<span style='color:orange'>

**SOMETHING TO REMEMBER:** Using Lambda Function

</span>

---

```python
new_df['above/below_avg'] = new_df.groupby('job')['outstanding_balance'].transform(lambda x: x - x.mean())
new_df
```

```python
new_df['avg_balance_with_loan'] = new_df[new_df['housing_loan'] == 'yes'].groupby('job')['outstanding_balance'].transform('mean')
new_df
```
