<div style='background-color:lightgray; padding:15px;'>
    
**`Understanding extract() Method`**

The `extract()` method is a powerful tool for extracting captured groups from strings within a Pandas Series or Index using regular expressions. It's particularly useful when you need to pull out specific parts of text based on a defined pattern.

**Key Features:**

  1. `Capturing Groups:` The core of `extract()` lies in its use of capturing groups within regular expressions. Capturing groups are defined by parentheses `(...)` in your regex pattern.  Each group captures a portion of the matched text.

  2. `DataFrame Output:` The extract() method returns a DataFrame, where each captured group becomes a separate column.

  3. `Handling Multiple Matches:` If a row matches multiple times, extract() returns NaN for that row.

</div>


```python
data = pd.Series(['apple-123', 'banana-456', 'cherry-789'])

# Extract the part after the hyphen
result1 = data.str.extract(r'-(\d+)')

print(result1)
```

```python
result2 = data.str.extract(r'\w+\-(\d+)')
result2
```

```python
result3 = data.str.extract(r'\-(\d+)')
result3
```

```python
data = pd.Series(['name=John,age=30', 
                  'name=Alice,age=25', 
                  'name=Bob,age=35'])
data
```

```python
# Extract name and age
result = data.str.extract(r'name=(\w+),age=(\d+)')

print(type(result))
result
```

```python
data = pd.Series(['name=John,age=30', 
                  'name=Alice,age=25', 
                  'name=Bob,age=35'])

# Extract name and age with named groups
result = data.str.extract(r'name=(?P<Name>\w+),age=(?P<Age>\d+)')

print(result)
```

# 10. PRACTICE LAB - Sandwich Generation

Yan is a sandwich enthusiast and is determined to try every combination of sandwich possible. He wants to start with every combination of bread and meats and then move on from there, but he wants to do it in a systematic way.

Below we have 2 tables, bread and meats

Output every possible combination of bread and meats to help Yan in his endeavors.

Order by the bread and then meat alphabetically. This is what Yan prefers.

```python
d1 = {'bread_id': [1, 2, 3, 4], 
      'bread_name': ['Whole Wheat', 'White', 
                     'Sourdough', 'Brioche']}
bread_table = pd.DataFrame(d1)

d2 = {
    'meat_id': [1, 2, 3, 4, 5, 6, 7],
    'meat_name':['Turkey', 'Ham', 'Roast Beef', 
                 'Pastrami', 'Salami', 
                 'Chicken', 'Bacon']}
meat_table = pd.DataFrame(d2)
```

```python
df1 = bread_table.assign(key=1)
df1
```

```python
df2 = meat_table.assign(key=1)
df2
```

```python
merged_df = pd.merge(df1, df2, on='key', how='inner')
merged_df
```

```python
# Trial No. 1
pd.merge(df1, df2, on='key', how='outer')
```

```python
# Trial No. 2
pd.merge(df1, df2, on='key', how='left')
```

```python
# Trial No. 3
pd.merge(df1, df2, on='key', how='right')
```

```python
# Trial No. 4
pd.merge(df1, df2, on='key', how='cross')
```

```python
result = merged_df[['bread_name', 'meat_name']].sort_values(by=['bread_name', 'meat_name'])
result
```
