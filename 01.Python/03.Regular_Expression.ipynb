## 1. Re Module + Regex Methods

</span>

---

`re module` provides functions and methods for working with regular expressions. Here is a list of the main regular expression methods available in the re module:


  * **<span style='color:red'>re.search(*pattern, string, flags=0*):</span>** Searches for a pattern anywhere in the string and returns a match object if found.

  * **<span style='color:red'>re.match(*pattern, string, flags=0*):</span>** Matches the pattern at the beginning of the string and returns a match object if found.

  * **<span style='color:red'>re.fullmatch(*pattern, string, flags=0*):</span>** Matches the entire string against the pattern and returns a match object if found.

  * **<span style='color:red'>re.findall(*pattern, string, flags=0*):</span>** Returns all non-overlapping occurrences of the pattern in the string as a list of strings.

  * **<span style='color:red'>re.finditer(*pattern, string, flags=0*):</span>** Returns an iterator yielding match objects for all non-overlapping occurrences of the pattern in the string.

  * **<span style='color:red'>re.split(*pattern, string, maxsplit=0, flags=0*):</span>** Splits the string by the occurrences of the pattern and returns a list of substrings.

  * **<span style='color:red'>re.sub(*pattern, repl, string, count=0, flags=0*):</span>** Replaces occurrences of the pattern in the string with the specified replacement string.

  * **<span style='color:red'>re.subn(*pattern, repl, string, count=0, flags=0*):</span>** Performs the same function as re.sub() but returns a tuple containing the modified string and the number of substitutions made.

  * **<span style='color:red'>re.compile(*pattern, flags=0*):</span>** Compiles a regular expression pattern into a pattern object, which can be used for matching or searching operations.

  * **<span style='color:red'>re.escape(*string*):</span>** Escapes special characters in the string, making it suitable for use as a literal in a regular expression pattern.

## 3. Regex Meta Characters

</span>

---

   * **<span style='color:orange'>[abc]:** Matches any single character that is either a, b, or c.

   * **<span style='color:orange'>[^abc]:** Matches any single character that is not a, b, or c. The ^ inside the square brackets negates the character class.

   * **<span style='color:orange'>[a-z]:** Matches any single lowercase letter from a to z.

   * **<span style='color:orange'>[A-Z]:** Matches any single uppercase letter from A to Z.

   * **<span style='color:orange'>[0-9]:** Matches any single digit from 0 to 9.

   * **<span style='color:orange'>[a-zA-Z]:** Matches any single letter, either lowercase or uppercase.

   * **<span style='color:orange'>[\d]:** Matches any single digit. It is equivalent to [0-9].

   * **<span style='color:orange'>[\D]:** Matches any single non-digit character. It is equivalent to [^0-9].

   * **<span style='color:orange'>[\w]:** Matches any single alphanumeric character (letters, digits, or underscore). It is equivalent to **`[a-zA-Z0-9_]`**.

   * **<span style='color:orange'>[\W]:** Matches any single non-alphanumeric character. It is equivalent to **`[^a-zA-Z0-9_]`**.

   * **<span style='color:orange'>[\s]:** Matches any whitespace character (space, tab, newline, etc.).

   * **<span style='color:orange'>[\S]:** Matches any non-whitespace character.

   * **<span style='color:orange'>[\b]:** Matches when there is a word boundary present (spaces mostly).

   * **<span style='color:orange'>[\B]:** Matches there is not a word boundary present (characters).

   * **<span style='color:orange'>[\Z]:** Matches at the end of a string

## 4. Match Object Methods

</span>

---

When using regular expressions with the re module, the re.match(), re.search(), and re.findall() functions return a match object that contains information about the pattern match. The match object provides various methods to access and manipulate the matched content. Here is a list of the main methods available on match objects:


  * **<span style='color:orange'>group():**</span> Returns the matched string.

  * **<span style='color:orange'>start():**</span> Returns the starting position of the match.

  * **<span style='color:orange'>end():**</span> Returns the ending position of the match.

  * **<span style='color:orange'>span():**</span> Returns a tuple containing the starting and ending positions of the match.

  * **<span style='color:orange'>groups():**</span> Returns a tuple containing all the captured groups in the match.

  * **<span style='color:orange'>group(index):**</span> Returns the captured group at the specified index.

  * **<span style='color:orange'>groupdict():**</span> Returns a dictionary containing named captured groups, where the keys are the group names.

  * **<span style='color:orange'>expand(template):**</span> Returns the matched string with any backreferences replaced using the specified template.

---

## 5. Regex Modifiers

</span>

---

Modifiers (*also known as flags*) are used to modify the behavior of the pattern matching. They are added as optional parameters to the regular expression functions or as inline modifiers within the pattern itself. Here is a list of the main modifiers in Python's re module:



* **<span style='color:orange'>re.IGNORECASE</span> - or - <span style='color:orange'>re.I:**</span> Ignores case sensitivity when matching characters. For example, /abc/i matches "abc", "AbC", "ABC", etc.

* **<span style='color:orange'>re.MULTILINE</span> - or - <span style='color:orange'>re.M:**</span>: Enables multi-line mode. Changes the behavior of ^ and $ to match the start and end of each line in addition to the start and end of the string.

* **<span style='color:orange'>re.DOTALL</span> - or - <span style='color:orange'>re.S:**</span> Makes the . metacharacter match any character, including newline characters (\n).

* **<span style='color:orange'>re.VERBOSE</span> - or - <span style='color:orange'>re.X:**</span> Allows the use of whitespace and comments within the regular expression pattern for better readability. Whitespace and comments are ignored unless they are within square brackets or escaped.

* **<span style='color:orange'>re.ASCII</span> - or - <span style='color:orange'>re.A:**</span> Restricts character interpretation to the ASCII character set. Overrides the use of Unicode for character classes like \w, \W, \b, etc.

* **<span style='color:orange'>re.UNICODE</span> - or - <span style='color:orange'>re.U:**</span> Enables Unicode matching. By default, \w, \W, \b, etc., use Unicode property escapes to match word characters and word boundaries.

* **<span style='color:orange'>re.DEBUG:</span>** Prints debugging information about the pattern compilation.

## 6.1. EXERCISE

</span>

---

Here we have a list of phone numbers. We want to extract all of the area codes into a list.

The area codes are the first 3 numbers in a phone number.

Write out the pattern needed and then create the for loop to insert the area codes into the "area_code_pattern" variable.

Starting code:

---

```python
# List of Phone Numbers
phone_numbers = ["(123) 456-7890", "(555) 123-4567", "(987) 654-3210", "(800) 555-1212", "(123) 456-7890"]

area_code_pattern = r'\((\d{3})'

# List to store extracted area codes:
area_codes = []

# Loop through each phone number and extract area codes:
for item in phone_numbers:
  code = re.findall(area_code_pattern, item)
  area_codes.append(code[0])

# Print out the list:
print(area_codes)
```
