### 4.1. Nested IF ELSE Statements

```python
grade = 99

if grade >= 90:
    if grade >= 97:
        print(str(grade) + " = A+")
    elif grade >= 93:
        print(str(grade) + " = A")
    else:
        print(str(grade) + " = A-")
elif grade >= 80:
    if grade >= 87:
        print(str(grade) + " = B+")
    elif grade >= 83:
        print(str(grade) + " = B")
    else:
        print(str(grade) + " = B-")
elif grade >= 70:
    if grade >= 77:
        print(str(grade) + " = C+")
    elif grade >= 73:
        print(str(grade) + " = C")
    else:
        print(str(grade) + " = C-")
elif grade >= 60:
    if grade >= 67:
        print(str(grade) + " = D+")
    elif grade >= 63:
        print(str(grade) + " = D")
    else:
        print(str(grade) + " = D-")
else:
    print(str(grade) + " = F")
```

### 4.3. Break, Continue, Else, Pass

```python
# Break - Example 1
counter = 0

while counter <= 10:
    print(counter)
    if counter == 6:
        break
    counter += 1
```

```python
# Break - Example 2
integers = [1, 2, 3, 4, 5, 6, 7]

for num in integers:
    power_of_num = num**num
    if power_of_num > 3000:
        break
    print(power_of_num)
```

```python
# Continue - Example 1
number = 0

while number < 5:
    number = number + 1
    if number == 3:
        continue
    print(number)
```

```python
# Continue - Example 2
integers = [1, 2, 3, 4, 5, 6, 7]

for num in integers:
    power_of_num = num**num
    if 200 <= power_of_num <= 3300:
        continue
    print(power_of_num)
```

```python
# Else - Example 1
counter = 0

while counter <=10:
    if counter == 6.5:
        break
    print(counter)
    counter += 1
else:
    print('No Breaks Found')
```

```python
# Else - Example 2
integers = [1, 2, 3, 4, 5]

for num in integers:
    power_of_num = num**num
    if power_of_num > 4000:
        break
    print(power_of_num)
else:
    power_of_num = num**num
    print(power_of_num+1)
```

```python
# Pass - Example 1
integers = [1, 2, 3, 4, 5]

for num in integers:
    pass
```

### 4.4. Nested While Loops

```python
number = 0
while number <= 5:
    print(number)
    inner_number = 0
    while inner_number < 3:
        print(inner_number)
        inner_number += 1
    number += 1
    print('---')
```

```python
number = 0
while number <= 5:
    print(number)
    inner_number = 0
    while inner_number < 3:
        inner_number += 1
        if inner_number == 2:
            continue
        print(inner_number)
    else:
        print('Inner While Loop Ended')
    number += 1
else:
    print('Outer While Loop Ended')
```

### 5.3. Lambda Function

</span>

Lambda functions, also known as anonymous functions, allow you to create small, one-line functions without a formal function definition. They are typically used for simple and short operations where a named function is not necessary. Lambda functions are defined using the lambda keyword followed by the function's parameters and a single expression.

```python
lambda x: x**2
```

```python
(lambda x: x**2)(6)
```

## 7.2. Try Except with Files

```python
path = 'Try-Except_Folder/MainFile.txt'

try:
    with open('Try-Except_Folder/NewData2.txt', 'r') as read_file:
        main_message = read_file.read()
        print('    - Variable Created! :)')
except Exception as e:
    print(e)
else:
    with open(path, 'a', newline='\n') as append_file:
        append_file.write(main_message)
finally:
    print("    ...Now we will print the Appended 'MainFile.txt'\n")
    with open(path, 'r') as read_file:
        print(read_file.read())
```
